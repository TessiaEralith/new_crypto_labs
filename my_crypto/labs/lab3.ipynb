Чтение с файла текста

```python

key_length = 15
with open('../files/7.15', 'r', encoding='IBM866') as f:
    text = f.readlines()

column = [[] for _ in range(key_length)]

for string in text:
    for i in range(key_length):
        column[i].append(string[i].lower())

table = [[] for _ in range(key_length)]
```

Анализ текста

```python
banned_bigram = [' ь', 'еь', 'ъь', 'ыь', 'аь', 'оь', 'эь', 'яь', 'иь', 'юь', 'ьь', '  ',
                 ' й', 'цй', 'кй', 'нй', 'гй', 'шй', 'щй', 'зй', 'хй', 'ъй', 'фй', 'вй', 'пй', 'рй', 'лй', 'дй', 'жй', 'чй', 'сй', 'мй', 'тй', 'ьй', 'бй', 'йй']
for i in range(key_length):
    for j in range(key_length):
        if i == j:
            table[i].append(0xffffff)
            continue
        num_banned_bigram = 0
        for letter_index in range(len(column[i])):
            bigram = column[i][letter_index] + column[j][letter_index]
            if bigram in banned_bigram:
                num_banned_bigram += 1
        table[i].append(num_banned_bigram)
print(table)
```

Преобразование результата

```python
test = {}
for j, i in enumerate(table):
    test[j] = i.index(min(i))

num_column = 0
for i in range(key_length):
    if i not in test.values():
        num_column = i
        break
key = []
for i in range(key_length):
    key.append(num_column)
    num_column = test[num_column]

decrypted_text = ""
for i in text:
    for j in key:
        decrypted_text += i[j]

print(decrypted_text.replace('_', ' '))
```
